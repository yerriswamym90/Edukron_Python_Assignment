```python
# Question 4: Find the cube root of 125
num = 125
cube_root = num ** (1/3)
print(cube_root)
```

```python
# Question 5: Calculate the sum of digits in number 12345
num = 12345
temp = num
sum = 0
while temp > 0:
    digit = temp % 10
    sum = sum + digit
    temp = temp // 10
print("Original Number:", num)    
print("sum of the num: ", sum)    
```

```python
# Question 6: Check if 97 is a prime number
num = 97
count = 0
for i in range(1, num + 1):
    if num % i == 0:
        count = count + 1
if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# Question 7: Find the factorial of 8
num = 8
fact = 1
for i in range(1,num+1):
    fact = fact * i
print(fact)    
```

```python
# Question 8: Calculate the average of numbers: 15, 23, 31, 42, 56
num = [15,23,31,42,56]
sum = 0
for i in num:
    sum = sum + i
average = sum / len(num)
print(average)
```

```python
# Question 9: Find the greatest common divisor (GCD) of 48 and 36
n1 = 48
n2 = 36
gcd = 1
for i in range(1, min(n1,n2)+1):
    if n1 % i == 0 and n2 % i == 0:
        gcd = i
print(gcd)        
```
