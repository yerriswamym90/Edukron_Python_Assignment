```python
# 4. Write a program that calculates BMI (Body Mass Index)
weight = float(input("Enter the weight (kg): "))
height = float(input("Enter the height (feet): "))/3.281
BMI = weight / (height ** 2)
print(f"The BMI is: {BMI}")

if BMI < 18.5:
    print("You are underweight.")
elif BMI >= 18.5 and BMI <= 24.9:
    print("You are Normal Weight")
elif BMI >= 25 and BMI <= 29.9:
    print("You are Overweight")
else:
    print("You are obese")
```

```python
# 1. Write a program that reverses the digits of a given number using a while loop.
#Example: Input: 12345, Output: 54321
#**Hint:** Initialize reverse=0. While n>0: get last digit (n%10), add to reverse (reverse=reverse*10+digit), remove last digit (n=n//10).
n = int(input("Enter the number"))
temp = n
reverse = 0
while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10
print("Original number:", n)
print("Revrsed number:", reverse)

```

```python
# 2. Write a program that calculates the sum of all digits in a given number.
#Example: Input: 12345, Output: 15 (1+2+3+4+5)
#**Hint:** While n>0: add last digit to sum (sum += n%10), then remove it (n = n//10).

n = int(input("Enter the number"))
sum = 0
temp = n
while temp > 0:
    digit = temp % 10
    sum = sum + digit
    temp = temp // 10
print(f"Sum of the {n} is {sum}")
```

```python
# 3. Write a number guessing game where:
# The program generates a random number between 1 and 100
# The user has to guess the number
# After each guess, the program tells the user if their guess is too high or too low
# The game continues until the user guesses correctly
# Display the number of attempts taken
#**Hint:** Use `import random` and `random.randint(1, 100)` to generate number. Use while loop with condition guess != target.
#Count attempts.

import random
target = random.randint(1, 100)
guess = None
attempts = 0
while guess != target:
    guess = int(input("Guess the number between 1 and 100: "))
    attempts += 1
    if guess < target:
        print("Too low! Try again.")
    elif guess > target:
        print("Too high! Try again.")
    else:
        print(f"Congratulations! You've guessed the number {target} in {attempts} attempts.")
```

```python
# 4. Write a program to check if a given number is a palindrome or not.
#A palindrome number reads the same forwards and backwards.Examples: 121, 12321, 1331 are palindromes
#**Hint:** Store original number, reverse it (like Problem 3.1), then compare original with reversed number.

n = int(input("Enter the number: "))
original = n
reversed_num = 0
while n > 0:
    reversed_num = reversed_num * 10 + n % 10
    n //= 10
if original == reversed_num:
    print(f"{original} is a palindrome.")
else:
    print(f"{original} is not a palindrome.")
```

```python
# 5. Write a program to check if a number is an Armstrong number (also called Narcissistic number).
#An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.
#Examples: - 153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 - 9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)
#**Hint:** First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

n = int(input("Enter the number:"))
sum = 0
temp = n
while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10
if n == sum:
    print(f"{n} is a Armstrong Number")
else:
    print(f"{n} is not a Armstrong number")
```

```python
# 1. Pattern Printing - Right Triangle (Easy)
# Print the following pattern for n rows:
# *
# **
# ***
# ****
# *****
# (Example shown for n=5)
# **Hint:** Use nested loops: outer loop for rows (i from 1 to n), inner loop for columns (j from 1 to i). Print * in inner loop.
n = int(input("Enter the number:"))
for i in range(1, n+1):
    for j in range(1, i+1):
        print("*", end =" ")
    print()
```

```python
# 2. Print the following pattern for n rows:
# *****
# ****
# ***
# **
# *
# (Example shown for n=5)
# **Hint:** Outer loop for rows. Inner loop prints stars from 1 to (n-i+1), where i is the row number.
n = int(input("Enter the number:"))
for i in range(1, n+1):
    for j in range(1, (n-i+2)):
        print("*", end = " ")
    print()
```

```python
# 3. Print the following pyramid pattern for n rows:
#     *
#    ***
#   *****
#  *******
# *********
# (Example shown for n=5)
# **Hint:** For row i: print (n-i) spaces, then print (2*i-1) stars. Use nested loops or string multiplication.
n = int(input("Enter the number:"))
for i in range(1, n+1):
    for j in range(n-i):
        print(" ", end = "")
    for k in range(2*i-1):
        print("*", end = "")
    print()
```

```python
# 5. Print a multiplication table grid from 1 to n.
# Example for n=5:
# 1   2   3   4   5
# 2   4   6   8  10
# 3   6   9  12  15
# 4   8  12  16  20
# 5  10  15  20  25
# **Hint:** Nested loops: for i in range(1, n+1) and for j in range(1, n+1). Print i*j with proper spacing.
n = int(input("Enter the number:"))
for i in range(1, n+1):
    for j in range(1, n+1):
        print(i * j, end = "  ")
    print()
```

```python
# 6. Print a diamond pattern for n rows (n should be odd):
#     *
#    ***
#   *****
#  *******
# *********
#  *******
#   *****
#    ***
#     *
# (Example shown for n=9)
# **Hint:** Print pyramid (upper half) for rows 1 to (n+1)//2, then print inverted pyramid (lower half) for remaining rows.
n = int(input("Enter the number:"))
for i in range(1, n+1):
    for j in range(n-i):
        print(" ", end = "")
    for k in range(2*i-1):
        print("*", end = "")
    print()
for i in range(1, n):
    for j in range(i):
        print(" ", end = "")
    for k in range(2*(n-i)-1):
        print("*", end = "")
    print()
```
