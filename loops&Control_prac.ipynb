```python
# 3. Write a number guessing game where:
# The program generates a random number between 1 and 100
# The user has to guess the number
# After each guess, the program tells the user if their guess is too high or too low
# The game continues until the user guesses correctly
# Display the number of attempts taken
#**Hint:** Use `import random` and `random.randint(1, 100)` to generate number. Use while loop with condition guess != target.
#Count attempts.

import random
target = random.randint(1, 100)
guess = None
attempts = 0
while guess != target:
    guess = int(input("Guess the number between 1 and 100: "))
    attempts += 1
    if guess < target:
        print("Too low! Try again.")
    elif guess > target:
        print("Too high! Try again.")
    else:
        print(f"Congratulations! You've guessed the number {target} in {attempts} attempts.")
```

```python
# 4. Write a program to check if a given number is a palindrome or not.
#A palindrome number reads the same forwards and backwards.Examples: 121, 12321, 1331 are palindromes
#**Hint:** Store original number, reverse it (like Problem 3.1), then compare original with reversed number.

n = int(input("Enter the number: "))
original = n
reversed_num = 0
while n > 0:
    reversed_num = reversed_num * 10 + n % 10
    n //= 10
if original == reversed_num:
    print(f"{original} is a palindrome.")
else:
    print(f"{original} is not a palindrome.")
```

```python
# 5. Write a program to check if a number is an Armstrong number (also called Narcissistic number).
#An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.
#Examples: - 153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 - 9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)
#**Hint:** First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

n = int(input("Enter the number:"))
sum = 0
temp = n
while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10
if n == sum:
    print(f"{n} is a Armstrong Number")
else:
    print(f"{n} is not a Armstrong number")
```

```python
# 2. Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.
# Algorithm: 1. Divide the larger number by the smaller number
# 2. Replace the larger number with the smaller number
# 3. Replace the smaller number with the remainder
# 4. Repeat until the remainder is 0
# 5. The GCD is the last non-zero remainder
# **Hint:** While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
while b != 0:
    rem = a % b
    a = b
    b = rem
print("GCD:", a)
```

```python
# 3. Write a program to check if a number is a perfect number.
# A perfect number is a positive integer that is equal to the sum of its proper positive divisors (excluding the number itself).
# Example: 6 is perfect because 6 = 1 + 2 + 3
# Example: 28 is perfect because 28 = 1 + 2 + 4 + 7 + 14
# **Hint:** Loop from 1 to n-1. If i divides n (n%i==0), add i to sum. Compare sum with n.
n = int(input("Enter the number: "))
sum = 0
if n > 1:
    for i in range(1, n):
        if n % i == 0:
            sum = sum + i
    if sum == n:    
        print(f"{n} is a perfect number.")
    else:
        print(f"{n} is not a perfect number.")
else:
    print("Enter a number greater than 1.")
```

```python
# 5. Write a program that converts a number (0-999) to words.
# - 0: Zero
# - 25: Twenty Five
# - 123: One Hundred Twenty Three
# - 500: Five Hundred
# **Hint:** Create lists for ones, teens, tens, etc. Extract hundreds, tens, ones digits. Convert each part to words and combine.
n = int(input("Enter a number (0-999): "))
if n < 0 or n > 999:
    print("Number out of range. Please enter a number between 0 and 999.")
else:
    ones = ["", "One", "Two", "Three", "Four", "Five", "Six", "Seven", "Eight", "Nine"]
    teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen", "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"]
    tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"]

    hundreds = n // 100
    tens_digit = (n % 100) // 10
    ones_digit = n % 10
    
    if hundreds > 0:
        result = ones[hundreds] + " Hundred"
    else:
        result = ""

    if tens_digit == 1:
        result += (" " + teens[ones_digit]) if result else teens[ones_digit]
    else:
        if tens_digit > 0:
            result += (" " + tens[tens_digit]) if result else tens[tens_digit]
        if ones_digit > 0:
            result += (" " + ones[ones_digit]) if result else ones[ones_digit]

    print("Number in words:", result)
```

```python
# 6. Write a program to check if a number is a Strong number.
# A Strong number is a number whose sum of factorial of digits is equal to the original number.
# Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145
# **Hint:** Extract each digit, calculate its factorial, add to sum. Compare sum with original number.
n = int(input("Enter the number: "))
original = n
for i in str(n):
    digit = int(i)
    factorial = 1
    for j in range(1, digit + 1):
        factorial *= j
    original -= factorial

if original == 0:
    print(f"{n} is a Strong number.")
else:
    print(f"{n} is not a Strong number.")
```

```python
# 7. Write a program to print Pascal's Triangle up to n rows.
# Example for n=5:
#     1
#    1 1
#   1 2 1
#  1 3 3 1
# 1 4 6 4 1
# Each number is the sum of the two numbers directly above it.
# **Hint:** Use formula: C(n,k) = C(n,k-1) * (n-k+1) / k, where C(n,0)=1. Or use nested loops with list to store previous row.
n = int(input("Enter the number of rows for Pascal's Triangle: "))
for i in range(n):
    print(" " * (n - i - 1), end="")
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print() 
```
