```python
# Question 12: Check if two strings are anagrams: "listen" and "silent"
str1 = "listen"
str2 = "silent"
count1 = {}
count2 = {}

for i in str1:
    if i in count1:
        count1[i] += 1
    else:
        count1[i] = 1

for i in str2:
    if i in count2:
        count2[i] += 1
    else:
        count2[i] = 1

if count1 == count2:
    print(f"{str1} and {str2} are Anagrams")
else:
    print(f"{str1} and {str2} are not Anagrams")
```

```python
# Question 42: Convert "hello world" to Morse code
str = "hello world"
morse = {
    'a': '.-',    'b': '-...',  'c': '-.-.',  'd': '-..',
    'e': '.',     'f': '..-.',  'g': '--.',   'h': '....',
    'i': '..',    'j': '.---',  'k': '-.-',   'l': '.-..',
    'm': '--',    'n': '-.',    'o': '---',   'p': '.--.',
    'q': '--.-',  'r': '.-.',   's': '...',   't': '-',
    'u': '..-',   'v': '...-',  'w': '.--',   'x': '-..-',
    'y': '-.--',  'z': '--..'
}

result = ""

for i in str:
    if i == " ":
        result += " / "
    else:
        result += morse[i] + " "

print(result)
```

```python
# Question 43: Find the longest common substring between "programming" and "grammar"
str1 = "programming"
str2 = "grammar"
longest = ""

for i in range(len(str1)):
    for j in range(len(str2)):
        k = 0

        while (i + k < len(str1) and j + k < len(str2) and str1[i + k] == str2[j + k]):
            k += 1

        if k > len(longest):
            longest = str1[i:i+k]

print(longest)
```

```python
# Question 47: Check if string is a valid IPv4 address: "192.168.1.1"
ip = "192.168.1.1"
parts = ip.split(".")
valid = True
if len(parts) != 4:
    valid = False
else:
    for i in parts:
        if not i.isdigit():
            valid = False
            break
        if int(i) < 0 or int(i) > 255:
            valid = False
            break
if valid:
    print("Valid IPv4 address")
else:
    print("Invalid IPv4 address")
```

```python
# Question 49: Convert "hello world" to ROT13 encoding
str1 = "hello world"
result = ""
for i in str1:
    if 'a' <= i <= 'z':
        result += chr((ord(i) - ord('a') + 13) % 26 + ord('a'))
    elif 'A' <= i <= 'Z':
        result += chr((ord(i) - ord('A') + 13) % 26 + ord('A'))
    else:
        result += i
print(result)
```

```python
# Question 50: Check if string is a valid credit card number: "4532015112830366"
card = "4532015112830366"
total = 0
reverse = card[::-1]
for i in range(len(reverse)):
    digit = int(reverse[i])
    if i % 2 == 1:
        digit = digit * 2
        if digit > 9:
            digit = digit - 9
    total = total + digit
if total % 10 == 0:
    print("Valid credit card number")
else:
    print("Invalid credit card number")
```
