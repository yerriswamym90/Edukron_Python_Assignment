```python
# Question 1: Create a set of first 10 natural numbers
s = set(range(1, 11))
print("Set of first 10 natural numbers:", s)
```

```python
# Question 2: Add element 11 to set {1, 2, 3, 4, 5}
s = {1, 2, 3, 4, 5}
s.add(11)
print("Set after adding 11:", s)
```

```python
# Question 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}
s = {1, 2, 3, 4, 5, 6}
s.remove(3)
print("Set after removing 3:", s)
```

```python
# Question 4: Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}
intersection = s1.intersection(s2)
print("Intersection:", intersection)
```

```python
# Question 5: Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}
difference = s1.difference(s2)
print("Difference:", difference)
```

```python
# Question 6: Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}
s = {1, 2, 3, 4, 6, 7, 8}
exists = 5 in s
print("Exists:", exists)
```

```python
# Question 7: Find the length of set {'a', 'b', 'c', 'd', 'e'}
s = {'a', 'b', 'c', 'd', 'e'}
length = len(s)
print("Length:", length)
```

```python
# Question 8: Create a set of vowels from string "hello world"
s = "hello world"
result = set()
for i in s.lower():
    if i in 'aeiou':
        result.add(i)
print("Set of vowels:", result)
```

```python
# Question 9: Remove duplicates from list [1, 2, 2, 3, 4, 4, 5, 6, 6, 7] using set
s = {1, 2, 2, 3, 4, 4, 5, 6, 6, 7}
unique_list = set(s)
print("List after removing duplicates:", unique_list)
```

```python
# Question 10: Check if {1, 2, 3} is a subset of {1, 2, 3, 4, 5, 6}
n = {1, 2, 3}
s = {1, 2, 3, 4, 5, 6}
is_subset = n.issubset(s)
print(f"Is {n} a subset of {s}? {is_subset}")
```
